## 1. Giới thiệu bài toán

Việc phân bổ thời gian hằng ngày luôn là một bài toán đau đầu với sinh viên. Ai cũng muốn điểm cao nhưng lại sợ bị quá tải và stress. Thực tế, chúng ta dành thời gian cho rất nhiều việc: tự học, ngủ, tham gia CLB, đi chơi với bạn bè hay tập thể dục. Nhưng câu hỏi đặt ra là: **Việc nào thực sự quyết định đến điểm GPA, và việc nào dễ làm chúng ta bị stress nhất?**

Thay vì phỏng đoán, nhóm quyết định sử dụng các mô hình Học máy (Machine Learning) để phân tích một bộ dữ liệu về thói quen sinh hoạt của sinh viên. Mục tiêu của dự án này là:
- **Mục tiêu 1:** Tìm ra thuật toán tốt nhất để dự đoán được điểm GPA và phân loại được mức độ Stress.
- **Mục tiêu 2:** Dùng mô hình đó để "đo lường" xem thói quen nào (Học, Ngủ, Chơi...) mới là yếu tố tác động mạnh nhất.
- **Mục tiêu 3:** Đưa ra một "thời gian biểu" tối ưu dựa trên số liệu thực tế để giúp các bạn sinh viên vừa học tốt vừa giữ được tâm lý thoải mái.

## 2. Mô tả dữ liệu

Bộ dữ liệu **Student Lifestyle Dataset** là một tập dữ liệu được thu thập nhằm phân tích mối quan hệ giữa lối sống và kết quả học tập của sinh viên. Dữ liệu bao gồm 2000 quan sát, với các biến phản ánh nhiều khía cạnh trong đời sống hàng ngày như thời gian học, hoạt động ngoại khóa, giấc ngủ, mức độ vận động thể chất, tương tác xã hội và mức độ căng thẳng, cùng với chỉ số kết quả học tập (CGPA). Cụ thể:

- `Student_ID`: mã của sinh viên;
- `Study_Hours_Per_Day`: trung bình số giờ học mỗi ngày;
- `Extracurricular_Hours_Per_Day`: trung bình số giờ hoạt động ngoại khóa mỗi ngày;
- `Sleep_Hours_Per_Day`: trung bình số giờ ngủ mỗi ngày;
- `Social_Hours_Per_Day`: trung bình số giờ giao lưu mỗi ngày;
- `Physical_Activity_Hours_Per_Day`: trung bình số giờ hoạt động thể chất mỗi ngày;
- `GPA`: điểm trung bình;
- `Stress_Level`: mức độ bị stress của sinh viên (Low, Moderate, High).

Bộ dữ liệu này cho phép nghiên cứu một cách hệ thống ảnh hưởng của các yếu tố hành vi và sức khỏe đến hiệu suất học tập, từ đó đặt ra các bài toán quan trọng như dự đoán kết quả học tập, phân tích yếu tố ảnh hưởng, và phân cụm các kiểu lối sống của sinh viên, nhằm hỗ trợ hiểu và cải thiện hiệu quả học tập trong bối cảnh giáo dục hiện đại.

Để giải quyết trọn vẹn yêu cầu, nhóm tiến hành xây dựng song song 2 luồng mô hình cho 2 bài toán riêng biệt. Tất cả các mô hình đều được tối ưu hóa siêu tham số tự động thông qua **Optuna** và **GridSearchCV**.

### 4.2. Bài toán Phân loại Stress (Classification)

- **Support Vector:** `SVC`

## 5. Đánh giá kết quả

### 5.1. Kết quả Regression (dự đoán GPA)

| Model | R² | MAE | RMSE |
|---|---|---|---|
| **LinearRegression** | **0.5494** | **0.1639** | **0.2051** |
| Ridge | 0.5493 | 0.1639 | 0.2051 |
| DecisionTreeRegressor | 0.5342 | 0.1671 | 0.2085 |
| SVR | 0.5005 | 0.1713 | 0.2160 |
| RandomForestRegressor | 0.4623 | 0.1792 | 0.2240 |

**Nhận xét**
- **Mô hình tốt nhất: `LinearRegression`** — R² cao nhất, MAE và RMSE thấp nhất trong nhóm.
- `Ridge` cho kết quả gần như giống `LinearRegression` (do chỉ thêm regularization nhẹ), xác nhận quan hệ giữa các biến thói quen và GPA mang tính **tuyến tính**.
- Các mô hình phức tạp hơn (`DecisionTreeRegressor`, `RandomForestRegressor`) cho kết quả **kém hơn** mô hình tuyến tính — dấu hiệu overfitting trên nhiễu khi quan hệ thực chất giữa input và GPA đơn giản, không cần mô hình phi tuyến phức tạp.
- R² toàn bộ chỉ đạt khoảng 0.50–0.55, cho thấy 5 biến thói quen hiện tại chỉ giải thích được khoảng một nửa phương sai của GPA — phần còn lại có thể do nhiễu hoặc thiếu các yếu tố quan trọng khác (năng lực nền, phương pháp học, môn học...).

### 5.2. Kết quả Classification (phân loại mức độ Stress)

| Model | Accuracy | Precision | Recall | F1-Score |
|---|---|---|---|---|
| LogisticRegression | 0.855 | 0.854 | 0.855 | 0.854 |
| KNeighborsClassifier | 0.880 | 0.879 | 0.880 | 0.879 |
| DecisionTreeClassifier | **1.000** | **1.000** | **1.000** | **1.000** |
| RandomForestClassifier | **1.000** | **1.000** | **1.000** | **1.000** |
| SVC | 0.955 | 0.955 | 0.955 | 0.955 |

**Nhận xét**
- Theo số liệu thuần túy, `DecisionTreeClassifier` và `RandomForestClassifier` đạt **accuracy tuyệt đối 1.0** — cao nhất trong bảng.
- **Tuy nhiên, kết quả này cần được xem xét cẩn trọng** trước khi kết luận là "tốt nhất". Độ chính xác 100% trên một bài toán dự đoán hành vi con người (stress) từ vài biến sinh hoạt là **bất thường**, thường là dấu hiệu của việc gán nhãn `Stress_Level` bằng công thức/ngưỡng cố định từ input.
- Nếu loại trừ 2 mô hình có kết quả đáng nghi này, **`SVC` (95.5%)** là lựa chọn đáng tin cậy nhất — hiệu suất cao nhưng vẫn nằm trong phạm vi hợp lý, không tuyệt đối.

**Phát hiện: `Stress_Level` mang tính rule-based**
Để kiểm chứng nghi vấn trên, nhóm đã thử giới hạn một cây quyết định chỉ được rẽ nhánh tối đa **3 lần** (`max_depth=3`) và huấn luyện trên toàn bộ dữ liệu. Mô hình tự học ra một luật rất đơn giản, đạt **accuracy = 1.0** ngay với luật này:

```
Study_Hours <= 8.05:
    Sleep_Hours <= 5.95   → High
    Sleep_Hours > 5.95:
        Study_Hours <= 5.95 → Low
        Study_Hours > 5.95  → Moderate
Study_Hours > 8.05 → High
```
Điều này cho thấy `Stress_Level` trong bộ dữ liệu được **xác định gần như tuyệt đối chỉ qua 2 ngưỡng đơn giản** trên `Study_Hours_Per_Day` và `Sleep_Hours_Per_Day` — đặc trưng của một bộ dữ liệu **tổng hợp (synthetic)**. Vì vậy nó không phản ánh đầy đủ độ phức tạp của stress trong đời thực.

**Nhận xét chi tiết:**

- Với GPA, 4 yếu tố còn lại (Ngủ, Vận động, Xã hội, Ngoại khóa) có mức độ ảnh hưởng khá đồng đều (~0.10), không có yếu tố phụ nào nổi bật hẳn lên để giành lấy vai trò thứ 2.

- **Kết luận:** Kết hợp với luật của cây quyết định ở mục 5.2 (chỉ cần `Study` và `Sleep` là phân loại được mức độ Stress), ta có thể khẳng định chắc chắn rằng: **Học tập** và **Giấc ngủ** là 2 biến số cốt lõi quyết định đến sự cân bằng của một sinh viên. 3 hoạt động còn lại chỉ đóng vai trò râu ria, thêm thắt.

## 6. Kết luận và hướng phát triển

Dựa trên các kết quả phân tích và đánh giá mô hình, nhóm đã tìm ra lời giải đầy đủ cho 3 mục tiêu đặt ra ở đầu bài báo cáo:

### 6.1. Mô hình tốt nhất
- **Đối với dự đoán GPA:** Mô hình **Linear Regression** là tốt nhất vì dữ liệu mang tính tuyến tính, khái quát hóa tốt mà không bị overfitting như các mô hình Tree-based.
- **Đối với phân loại Stress:** Mô hình **SVC** là sự lựa chọn đáng tin cậy nhất (Accuracy 95.5%), hợp lý về mặt thống kê và thực tiễn hơn so với kết quả tuyệt đối bất thường của Decision Tree (do tính chất nhân tạo của dữ liệu).

### 6.2. Thói quen tác động mạnh nhất
- Điểm số GPA hoàn toàn bị chi phối bởi **thời gian tự học (Study_Hours)**.
- Mức độ Stress bị chi phối mạnh mẽ nhất bởi sự kết hợp giữa **thời gian tự học (Study_Hours)** và **thời gian ngủ (Sleep_Hours)**.

### 6.3. Thời gian biểu tối ưu
Chính vì Tự học và Giấc ngủ là 2 yếu tố then chốt, giải pháp cốt lõi để sinh viên đạt GPA tối ưu mà KHÔNG bị stress là phải giữ 2 biến số này trong "vùng an toàn". Dưới đây là khuyến nghị chi tiết (được rút ra từ nhóm top 25% sinh viên có GPA ≥ 3.30 và không bị Stress).

| Hoạt động | Khoảng khuyến nghị (giờ/ngày) |
|---|---|
| **Tự học** | 7.1 – 7.8 |
| **Ngủ** | 6.8 – 8.5 |
| **Giao tiếp xã hội** | 1.4 – 4.2 |
| **Ngoại khóa** | 0.9 – 2.8 |
| **Vận động thể chất** | 2.2 – 6.0 |

### 6.4. Hướng phát triển
Bộ dữ liệu hiện tại có dấu hiệu tổng hợp (synthetic) cho nhãn Stress_Level. Hướng phát triển trong tương lai là thu thập dữ liệu thực tế (real-world dataset) để mô hình có thể nắm bắt chính xác hơn sự phức tạp của tâm lý và sinh hoạt con người, đồng thời bổ sung thêm các yếu tố nền tảng (năng lực, áp lực tài chính, phương pháp học).
